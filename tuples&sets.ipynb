```python
s1={1,2,3,4}
s2={3,4,6,7,89}
print(s1|s2)    #UNION
print(s1&s2)    #INTERSECTION
print(s1-s2)    #DIFFERENCE
print(s1^s2)    #SYMMETRIC DIFFERENCE
```

```python
s1={3,4,6,7,89,"chandu",True}
print(s1)
s1.add(250)
print(s1)
s2={(2,3),345,"hi"}
s1.update(s2)
print(s1)
s1.remove("hi")
print(s1)
s1.discard(200)
print(s1)
print(s1.pop())
print(s1)
s2.clear()
print(s2)
print(s1.union(s2))
```
