```python
# 1. Creating a dictionary
d = {
    "name": "sai",
    "age": 20,
    "grade": 9.3,
    "height": 6.0,
    "NAME": "sri"
}
print(d)
```

```python
d = {}
print(type(d))
```

```python
d1 = dict(a=70, n=34)
print(d1)
```

```python
for i in d:
    print(i)
```

```python
x = d.keys()
print(x)
print(type(x))
print(list(x))
```

```python
y = d.values()
print(y)
print(list(y))
```
